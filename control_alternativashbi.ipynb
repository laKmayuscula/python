```python
#ejc 7
base = input("Dime la base: ")
exponente = input("Dime el exponente: ")
resultado = int(base) ** int(exponente)

if int(exponente) == 0:
    print("El resultado de la potencia es: 1")
else:
    print("el resultado de la potencia es: ", resultado)
```

```python
#ejc 8
nota = int(input("Dime la nota: "))
edad = int(input("Dime la edad: "))
sexo = input("Dime el sexo: (M/F) ")

if nota >= 5 and edad >= 18 and sexo == "F":
    print("ACEPTADA")

elif nota >= 5 and edad >= 18 and sexo == "M":
    print("POSIBLE")

else:
    print("NO ACEPTADA")
```
